## Cleaning the Data Set

#### Nan value plotting

```python
import matplotlib.pyplot as plt
plt.bar(df.columns, df.isna().sum())
plt.xticks(rotation=90)
plt.show()
```

#### Finding Total percentage of Nan values

```python
nan_col = df.columns[df.isna().any()]
nan_p_df = pd.DataFrame({'Columns': nan_col, 
                        'Nan Percentage': [round((df[i].isna().sum()/df.shape[0])* 100, 1) for i in nan_col]})
```

```python
nan_p_df
```

#### Converting the Age columns into integers

```python
s_index = []
for i in range(len(df.age)):
    if isinstance(df.age[i], str):
        s_index.append(i)

for i in s_index:
    df.age[i] = df.age[i][:2]
df.age = df.age.astype(int)
df.age.dtype
```

```python
set(df.gender)
```

```python
df.gender =df.gender.replace(['F', 'FEMALE', 'female'], 'Female')
df.gender = df.gender.replace(['M', 'MALE', 'male'], 'Male')
set(df.gender)
```
